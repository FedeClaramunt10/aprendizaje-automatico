## ¿Qué problema vamos a resolver?

Buscamos identificar grupos de futbolistas con características similares utilizando técnicas de aprendizaje no supervisado. No contamos con etiquetas previas, por lo que los algoritmos deberán descubrir patrones automáticamente.

Cargamos los datos bajados de la database de [Kaggle](https://www.kaggle.com/karangadiya/fifa19) como 'data.csv'.

````python
df=pd.read_csv(url)
````

## Exploración inicial

Antes de aplicar clustering es importante comprender los datos: cantidad de jugadores, variables disponibles, calidad de los datos y posibles valores faltantes.

**Preparamos la base para Clustering:**

**Retiramos jugadores con bajo Desempeño global ("Overall")**

**y consideramos Variables de desempeño por habilidad (skills_ratings) como precisión de pases, velocidad, etc.
Esto es lo que definirá nuestro "espacio" de trabajo.**

````python
n=10000
df_n=df.loc[:n] #los primeros n
df_n=df_n[(df_n['Overall']>70)] # mejores jugadores

skills_ratings = ['Crossing', 'Finishing', 'HeadingAccuracy', 'ShortPassing',
                  'Volleys', 'Dribbling', 'Curve', 'FKAccuracy', 'LongPassing',
                  'BallControl', 'Acceleration', 'SprintSpeed', 'Agility', 'Reactions',
                  'Balance', 'ShotPower', 'Jumping', 'Stamina', 'Strength', 'LongShots',
                  'Aggression', 'Interceptions', 'Positioning', 'Vision', 'Penalties',
                  'Composure', 'Marking', 'StandingTackle', 'SlidingTackle',
                  'GKDiving', 'GKHandling', 'GKKicking', 'GKPositioning', 'GKReflexes']

print(len(skills_ratings), 'variables numéricas: skills/habilidades')

df_skills=df_n[skills_ratings]
````

````python
df_skills #Entrada, matriz de datos X
````

## Representación de los jugadores

Cada jugador será representado mediante un vector de habilidades (velocidad, remate, pase, defensa, etc.). Los algoritmos agruparán jugadores con vectores similares.

## K-Means: idea general

1. Selecciona K centroides iniciales.
2. Asigna cada observación al centroide más cercano.
3. Recalcula los centroides.
4. Repite hasta converger.

Objetivo: minimizar la distancia entre los puntos y el centroide de su grupo.

# 3.-Clustering-con-Kmeans

Es necesario definirle la cantidad de grupos/clusters (n_clusters), es un hiperparámetro del algoritmo.


````python
# Número de clusters buscado
n_clust = 5

km = KMeans(n_clusters=n_clust)
km.fit(df_skills) #utiliza todas las habilidades: 34 dimensiones

# Etiquetas asignadas por el algoritmo
clusters = km.labels_
print('Suma de los cuadrados de las distancias al centro de cada cluster=Inertia= ', km.inertia_)
````

La Inertia mide cuánto se parecen los integrantes de cada grupo entre sí. Si todos los jugadores de un cluster tienen características muy similares, la Inertia será pequeña. Si mezclamos delanteros, defensores y arqueros en un mismo grupo, la Inertia aumentará porque los jugadores estarán más alejados de su centroide.

````python
df_clusters=df_n.copy()
df_clusters['kmeans_5'] = km.labels_ #clusters
print('Kmeans encontró: ', max(km.labels_)+1, 'clusters, nosotros forzamos la cantidad')
df_clusters.head(4)
````

**Graficamos el resultado de agrupar a los jugadores usando kmedias con la cantidad de clases elegida: Visualizamos de a dos variables por vez**

````python
bool_crack=df_n["Overall"] > 85
bool_no_crack=df_n["Overall"]<86

#Eligo dos números entre 0 y  de n_skills-1
skill_1=skills_ratings[9]
skill_2=skills_ratings[10]
````

````python
kmean_clusters = go.Scatter(x=df_skills[skill_1], y=df_skills[skill_2],
                           mode='markers',
                        text=df_n.loc[:,'Name'],
                           marker=dict(
                                size=5,
                                color = clusters.astype(float), #set color equal to a variable
                                colorscale='Portland',
                                showscale=False)
                           )

crack =go.Scatter(x=df_skills.loc[bool_crack,skill_1], y=df_skills.loc[bool_crack,skill_2],name='Ckacks!!',
                      text=df_n.loc[bool_crack,'Name'],
                      textfont=dict(family='sans serif',size=10,color='black'),
                      opacity=0.9,mode='text')

data=[kmean_clusters,crack]

layout = go.Layout(title="Clustering K means ",titlefont=dict(size=20),
                xaxis=dict(title=skill_1),
                yaxis=dict(title=skill_2),
                autosize=False, width=1000,height=650)

fig = go.Figure(data=data, layout=layout)
````

````python
#En Colab:
fig.show(renderer="colab")
````

#Volvemos a la Teoría

Ayuda: Creamos una nueva variable categórica de la base recodificando la posición para simplificar el análisis de los resultados de clustering

````python
forwards=['RF', 'ST', 'LW', 'LF', 'RS', 'LS', 'RM', 'LM','RW']
midfielders=['RCM','LCM','LDM','CAM','CDM','LAM','RDM','CM','RAM','CF']
defenders=['RCB','CB','LCB','LB','RB','RWB','LWB']
goalkeepers=['GK']

def pos2(position):
    if position in forwards:
        return 'Forward'

    elif position in midfielders:
        return 'Midfielder'

    elif position in defenders:
        return 'Defender'

    elif position in goalkeepers:
        return 'GK'

    else:
        return 'nan'
````

````python
df_n["Position2"]=df_n["Position"].apply(lambda x: pos2(x))

df_n["Position2"].value_counts()
````

**Elección de la cantidad de grupos/clusters**

Una forma de seleccionar la cantidad más adecuada de clusters es probar varias cantidades y usar el criterio del codo: "elbow method", utilizando una medida de ajuste que puede ser Inertia. Usamos la medida de Inercia para encontrar el número óptimo de clusters, buscando el punto donde la ganancia de ajuste deja de ser significativa.

````python
#Prueba: para elegir el hiperparámetro n_clusters, variando de 2 a 11 clusters
scores = [KMeans(n_clusters=i).fit(df_skills).inertia_ for i in range(2,12)]

plt.plot(np.arange(2, 12), scores)
plt.xlabel('Number of clusters')
plt.ylabel("Inertia")
plt.title("Inertia of k-Means versus number of clusters")
````

# 4.-Clustering-Mean-Shift

- NO es necesario definirle la cantidad de grupos/clusters
- NO es escalable, computacionalmente es muy pesado. Cada dato tienen un proceso iterativo en donde en cada iteración considera los vecinos a menos de un radio de distancia.
- Hipeparámetros: radio/bandwith (de cobertura para cálculo de media). (Ejercicio: Probar con varios)

````python
ms = MeanShift(bandwidth=2, bin_seeding=True)

ms.fit(df_skills)

clusters2 = ms.labels_
cluster_centers = ms.cluster_centers_

labels_unique = np.unique(clusters2)
n_clusters_ = len(labels_unique)

print("Cantidad de clusters encontrados por Mean Shift : %d" % n_clusters_)
````

Graficamos los resultados:

````python
bool_crack=df_n["Overall"] > 85
bool_no_crack=df_n["Overall"]<86

#Eligo dos números entre 0 y  de n_skills-1
skill_1=skills_ratings[5]
skill_2=skills_ratings[14]
````

````python
data=[]
MeanShift_clusters = go.Scatter(x=df_skills[skill_1], y=df_skills[skill_2],
                           mode='markers',
                        text=df_n.loc[:,'Name'],
                           marker=dict(
                                size=5,
                                color = clusters2.astype(float), #set color equal to a variable
                                colorscale='Portland',
                                showscale=False)
                           )

crack =go.Scatter(x=df_skills.loc[bool_crack,skill_1], y=df_skills.loc[bool_crack,skill_2],name='Top players',
                      text=df_n.loc[bool_crack,'Name'],
                      textfont=dict(family='sans serif',size=10,color='black'),
                      opacity=0.9,mode='text')

data=[MeanShift_clusters,crack]

layout = go.Layout(title="Mean Shift",titlefont=dict(size=20),
                xaxis=dict(title=skill_1),
                yaxis=dict(title=skill_2),
                autosize=False, width=1000,height=650)

fig = go.Figure(data=data, layout=layout)
````

````python
from sklearn.cluster import MeanShift

ms1 = MeanShift(bandwidth=1)
clusters1 = ms1.fit_predict(X_scaled)

print("Cantidad de clusters:", len(set(clusters1)))
````

Bandwidth = 3

````python
ms3 = MeanShift(bandwidth=3)
clusters3 = ms3.fit_predict(X_scaled)

print("Cantidad de clusters:", len(set(clusters3)))
````

Bandwidth = 5

````python
ms5 = MeanShift(bandwidth=5)
clusters5 = ms5.fit_predict(X_scaled)

print("Cantidad de clusters:", len(set(clusters5)))
````

3.a)_ Con un bandwidth de 1, el algoritmo encuentra 2 clusters.
Al aumentar el bandwidth a 3 y 5, el algoritmo colapsa toda la estructura en 1 solo cluster.

3.b)_ La configuración más razonable es bandwidth = 1.

El propósito del clustering es identificar subgrupos o patrones distintos dentro de un conjunto de datos. Cuando el bandwidth es demasiado alto (como 3 o 5), el algoritmo tiene un radio de búsqueda tan grande que considera que todos los puntos pertenecen a una sola densidad, perdiendo información valiosa sobre la estructura subyacente. Un bandwidth de 1 permite que el algoritmo sea más sensible a las variaciones locales y logre separar los datos en grupos significativos.
